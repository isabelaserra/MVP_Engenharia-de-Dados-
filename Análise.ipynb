```python
%sql
USE workspace.consumidor;

-- As 10 empresas com mais reclamações e sua participação no total
SELECT
  dim_empresa.nome_fantasia,
  COUNT(*) AS total_reclamacoes,
  ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 1) AS percentual
FROM fato_reclamacao
JOIN dim_empresa ON fato_reclamacao.id_empresa = dim_empresa.id_empresa
GROUP BY dim_empresa.nome_fantasia
ORDER BY total_reclamacoes DESC
LIMIT 10;
```

```python
%sql
USE workspace.consumidor;

SELECT
  dim_problema.assunto,
  dim_problema.problema,
  COUNT(*) AS total_reclamacoes
FROM fato_reclamacao
JOIN dim_problema ON fato_reclamacao.id_problema = dim_problema.id_problema
GROUP BY dim_problema.assunto, dim_problema.problema
ORDER BY total_reclamacoes DESC
LIMIT 10;
```

```python
%sql
USE workspace.consumidor;

SELECT
  ROUND(AVG(tempo_resposta), 1) AS tempo_medio_dias,
  ROUND(AVG(nota_consumidor), 2) AS nota_media
FROM fato_reclamacao;
```

```python
%sql
USE workspace.consumidor;

SELECT
  CASE
    WHEN tempo_resposta <= 3 THEN '1. Até 3 dias'
    WHEN tempo_resposta <= 7 THEN '2. De 4 a 7 dias'
    ELSE '3. Mais de 7 dias'
  END AS faixa_tempo,
  COUNT(*) AS reclamacoes,
  ROUND(AVG(nota_consumidor), 2) AS nota_media
FROM fato_reclamacao
WHERE tempo_resposta IS NOT NULL
GROUP BY faixa_tempo
ORDER BY faixa_tempo;
```

```python
%sql
USE workspace.consumidor;

SELECT
  avaliacao_reclamacao,
  COUNT(*) AS total,
  ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 1) AS percentual
FROM fato_reclamacao
GROUP BY avaliacao_reclamacao
ORDER BY total DESC;
```

```python
%sql
USE workspace.consumidor;

SELECT
  avaliacao_reclamacao,
  COUNT(*) AS total,
  ROUND(AVG(nota_consumidor), 2) AS nota_media
FROM fato_reclamacao
GROUP BY avaliacao_reclamacao
ORDER BY nota_media DESC;
```

```python
%sql
USE workspace.consumidor;

SELECT
  dim_local.regiao,
  COUNT(*) AS total_reclamacoes,
  ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 1) AS percentual
FROM fato_reclamacao
JOIN dim_local ON fato_reclamacao.id_local = dim_local.id_local
GROUP BY dim_local.regiao
ORDER BY total_reclamacoes DESC;
```
